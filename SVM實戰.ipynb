## Step1. 下載資料

這是一組常用於機器學習（特別是乳腺癌診斷資料集，如威斯康辛大學的 Breast Cancer Wisconsin Dataset）的醫學特徵欄位名稱 [1, 2]。以下為你提供精準的專業繁體中文翻譯：mean radius：半徑平均值mean texture：紋理平均值mean perimeter：周長平均值mean area：面積平均值mean smoothness：平滑度平均值mean compactness：緊湊度（緻密度）平均值mean concavity：凹度平均值mean concave points：凹點數平均值mean symmetry：對稱性平均值mean fractal dimension：分形維度（斷形維度）平均值

````python
breast_cancer=datasets.load_breast_cancer()

breast_cancer_df = pd.DataFrame(
    breast_cancer.data,
    columns=breast_cancer.feature_names
)
breast_cancer_df.head(10)

````

````python
X=breast_cancer.data

#0 = malignant 惡性
#1 = benign 良性
y=breast_cancer.target
````

## Step2. 區分訓練集與測試集

````python
X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.3,random_state=0)
````

## Step3. 建模

````python
#clf 是 classifier 的縮寫，意思是「分類器」。
#使用 sklearn 裡的 Support Vector Classification，也就是 SVM 分類模型。
#使用「多項式核函數」。
#當資料不是直線可以分開時，SVM 會透過 kernel 把資料轉換到較高維度，讓它比較容易被分開。
#gamma = 1 / 特徵數量 
#gamma 越大，每個點影響範圍越小，模型可能越複雜；gamma 越小，每個點影響範圍越大，模型會比較平滑。
#C 越大，模型越不允許分錯，會努力把訓練資料分得很準，但比較容易 overfitting。
#C 越小，模型比較能容忍一些錯誤，邊界會比較平滑，但可能 underfitting。

clf=svm.SVC(kernel='poly',gamma='scale',C=100)
clf.fit(X_train,y_train)
````

## Step4. 預測

## Step5. 準確度分析

````python
print(clf.score(X_train,y_train))
print(clf.score(X_test, y_test))
````

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets, svm
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

breast_cancer = datasets.load_breast_cancer()

X = breast_cancer.data[:, [0, 1]]
y = breast_cancer.target

clf = make_pipeline(
    StandardScaler(),
    svm.SVC(kernel='poly', gamma='scale', C=1)
)

clf.fit(X, y)

def plot_svm_2d(clf, X, y, title='SVM prediction result'):
    x_min = X[:, 0].min() - 1
    x_max = X[:, 0].max() + 1
    y_min = X[:, 1].min() - 1
    y_max = X[:, 1].max() + 1

    xx, yy = np.meshgrid(
        np.linspace(x_min, x_max, 500),
        np.linspace(y_min, y_max, 500)
    )

    grid_points = np.c_[xx.ravel(), yy.ravel()]
    Z = clf.predict(grid_points)
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(8, 6))
    plt.contourf(xx, yy, Z, alpha=0.25, cmap=plt.cm.coolwarm)
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.coolwarm, edgecolor='k', s=35)

    plt.xlabel(breast_cancer.feature_names[0])
    plt.ylabel(breast_cancer.feature_names[1])
    plt.title(title)
    plt.show()


plot_svm_2d(clf, X, y, 'SVM poly prediction result')
````
